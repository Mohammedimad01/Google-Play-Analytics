```python
from scipy import stats

free_ratings = apps[apps['Type'] == 'Free']['Rating'].dropna()
paid_ratings = apps[apps['Type'] == 'Paid']['Rating'].dropna()

print(f"Free apps: n={len(free_ratings)}, mean={free_ratings.mean():.3f}")
print(f"Paid apps: n={len(paid_ratings)}, mean={paid_ratings.mean():.3f}")

t_stat, p_value = stats.ttest_ind(free_ratings, paid_ratings, equal_var=False)
print(f"t-statistic: {t_stat:.3f}, p-value: {p_value:.5f}")
```

```python
from scipy.stats import pearsonr

for col in ['Size_MB', 'Price', 'Reviews']:
    valid = apps[[col, 'Rating']].dropna()
    r, p = pearsonr(valid[col], valid['Rating'])
    print(f"{col} vs Rating: r={r:.3f}, p={p:.5f}")
```

```python
import statsmodels.api as sm

reg_data = apps[['Rating', 'Price', 'Size_MB', 'Reviews', 'Installs']].dropna()
X = reg_data[['Price', 'Size_MB', 'Reviews', 'Installs']]
X = sm.add_constant(X)
y = reg_data['Rating']

model = sm.OLS(y, X).fit()
print(model.summary())
```

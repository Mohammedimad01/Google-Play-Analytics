```python
import pandas as pd

apps_url = "https://raw.githubusercontent.com/malborroni/Foundations_of_Computer-Science/master/datasets/googleplaystore.csv"
reviews_url = "https://raw.githubusercontent.com/Ciroye/sentiment-analysis-google-play-reviews/master/googleplaystore_user_reviews.csv"

apps = pd.read_csv(apps_url)
reviews = pd.read_csv(reviews_url)

print(apps.shape)
print(reviews.shape)
```

```python
print(apps.info())
print(apps.isnull().sum())
print("Duplicate rows:", apps.duplicated().sum())
print(apps['Category'].unique())
```

```python

apps = apps[apps['Category'] != '1.9'].reset_index(drop=True)
```

```python
apps['Reviews'] = apps['Reviews'].astype(int)
```

```python
def parse_size(x):
    if x == 'Varies with device':
        return None
    if x.endswith('M'):
        return float(x[:-1])
    if x.endswith('k'):
        return float(x[:-1]) / 1024   # k → M
    return None

apps['Size_MB'] = apps['Size'].apply(parse_size)
```

```python
apps['Installs'] = apps['Installs'].str.replace(',', '').str.replace('+', '').astype(int)
```

```python
apps['Type'] = apps['Type'].fillna(apps['Price'].apply(lambda p: 'Free' if p == '0' else 'Paid'))
```

```python
apps['Price'] = apps['Price'].str.replace('$', '').astype(float)
```

```python
apps['Last Updated'] = pd.to_datetime(apps['Last Updated'])
```

```python
print(f"Missing ratings: {apps['Rating'].isnull().sum()}")
```

```python
apps = apps.drop_duplicates(subset='App', keep='first').reset_index(drop=True)
print(apps.shape)
```

```python
reviews_clean = reviews.dropna(subset=['Translated_Review']).reset_index(drop=True)
print(reviews.shape, "→", reviews_clean.shape)
```

```python
apps.to_csv('googleplaystore_clean.csv', index=False)
reviews_clean.to_csv('googleplaystore_reviews_clean.csv', index=False)

from google.colab import files
files.download('googleplaystore_clean.csv')
files.download('googleplaystore_reviews_clean.csv')
```
